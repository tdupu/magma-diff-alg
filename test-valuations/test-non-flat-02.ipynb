```magma
AttachSpec("diffalg.spec");
```

```magma
Z:=Integers();
Q:=RationalField();
P<x,y>:=PolynomialRingProlSeq(Q,2: term_order:=<"dblocks",[[1,2]]>);
```

```magma
u:=y+Diff(x,1)+Diff(x,2);
v:=y+Diff(x,1)+Diff(x,3);
gens:=[u,v];
ders:=Jet([Diff(f,r) : r in [0..5], f in gens]);
R8:=Jet(P,8);
```

```magma
ders[1] in Jet(P,8);
I8:=ideal<R8|ders>;
```

```magma
Cs,Is:=CharacteristicSeries([u,v]);
sys:=Cs[1];
eqns:=sys[1];
ineqns:=sys[2];
pp1:=JetIdeal(eqns,ineqns);
```

```magma
Dimension(pp1);
A:=DifferentialRing(eqns,ineqns);
```

```magma
ord:=8;
elim_vars:={Jet(Diff(v,r),ord) : r in [0..2], v in {x,y}};
```

```magma
pp2:=EliminationIdeal(I8,elim_vars);
pp2;
pp1;
```

```magma
P2<x,y>:=PolynomialRingProlSeq(Q,2: term_order:=<"dblocks",[[1],[2]]>);
R8_2:=Jet(P2,8);
ChangeRing(I8,R8_2);
```

```magma
R8_Q:=ChangeRing(R8,Q);
R8_2_Q:=ChangeRing(R8_2,Q);
```
