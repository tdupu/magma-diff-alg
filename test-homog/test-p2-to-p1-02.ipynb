```magma
AttachSpec("../diffalg.spec");
Z:=Integers();
Q:=RationalField();

/*
Find order 2 degree 3 elements, 
dehomogenize and determine if they generate a unit ideal. 
*/
```

```magma
R<t>:=PolynomialRing(Q,1);
f := map<R->R|f:->Derivative(f,t)>;
A<t> := DifferentialRing(R, f, Q);
F<t>:=FieldOfFractions(A);
P<x0,x1,x2>:=PolynomialRingProlSeq(F,3: term_order:=<"dblocks",[[1],[2],[3]]>);
```

```magma
gens:=WronskianGenerators(3,[x0,x1,x2]);
#gens;
last:=[gens[i] : i in [#gens-27..#gens]];
```

```magma
/*
This shows that pairwise every two have some base loci.
*/
for i in [1..#last] do
    for j in [i+1..#last] do
        A:=last[i];
        B:=last[j];
        seq:=[A,B];
        seq0:=[Evaluate(f,[P!1,x1,x2]) : f in seq];
        Ch:=CharacteristicSeries(seq0);
        for ch in Ch do
            eqns:=ch[1];
            Quo,Rem,Sep:=RittDivide(P!1,eqns);
            if Rem eq 0 then
                i,j,ch;
            else
                //print "failed\n";
            end if;
        end for;
    end for;
end for;
```
